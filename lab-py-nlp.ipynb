## **2. Text Preprocessing**

### **Exercise 1: Tokenization and Stop Word Removal**

Tokenize the following text

```python
# ------------------------------------------------------------------ #
#                 Tokenization de chaîne de caractère                #
# ------------------------------------------------------------------ #

from nltk.tokenize import word_tokenize # Import

text = "Natural Language Processing (NLP) is a fascinating field of study! It involves analyzing and understanding human language."
tokenized_text = word_tokenize(text=text, language="english", preserve_line=False)

# preserve_line=False pour découper en phrases propres. ex. Texte long, multi-paragraphes
# preserve_line=True pour garder la structure sans segmentation automatique. ex. Prétraitement brut de données avant NLP avancé

print(tokenized_text)
```

Remove stop words and store the result in a variable called `filtered_tokens`

```python
# ------------------------------------------------------------------ #
#                    Mots de liaison + ponctuation                   #
# ------------------------------------------------------------------ #

# Import de la liste des stopwords en anglais
from nltk.corpus import stopwords

# Définir la liste des stopwords anglais une fois
stop_words = set(stopwords.words('english'))

# Conversion en lowercase + removal ponctuation
def to_lowercase_no_punct(tokens):
  return [word.lower() for word in tokens if word.isalpha()]

# Fonction pour enlever les stopwords sans perdre l’ordre
def remove_sw(tokens):
    return [word for word in tokens if word.lower() not in stop_words]

# Application sur tokenized_text
tokenized_text_to_lowercase = to_lowercase_no_punct(tokenized_text)
filtered_tokens = remove_sw(tokenized_text_to_lowercase)

print(filtered_tokens)
```

### **Exercise 2: Stemming and Lemmatization**

Apply stemming and lemmatization to the `filtered_tokens`. Compare the results.

Apply stemming and store the result in `stemmed_tokens`

```python
# ------------------------------------------------------------------ #
#                            Stemming data                           #
# ------------------------------------------------------------------ #

# Import stemmer
from nltk.stem import PorterStemmer

# Instanciate
stemmer = PorterStemmer()

# Fonction pour stemmatiser
def stemmatize(tokens):
    return [stemmer.stem(word) for word in tokens]

# Application
stemmed_tokens = stemmatize(filtered_tokens)

print("Stemmed Tokens:", stemmed_tokens)
```

Apply lemmatization and store the result in `lemmatized_tokens`

```python
# ------------------------------------------------------------------ #
#                          Lemmatizing data                          #
# ------------------------------------------------------------------ #

# Import stemmer
from nltk.stem import WordNetLemmatizer

# Instanciate
lemmatizer = WordNetLemmatizer()

# Fonction pour stemmatiser
def lemmatize(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]

# Application
lemmatized_tokens = lemmatize(filtered_tokens)

print("Lemmatized Tokens:", lemmatized_tokens)
```

## **3. Feature Extraction**

### **Exercise 3: Bag of Words (BoW)**

Use the `CountVectorizer` from `scikit-learn` to create a Bag of Words representation of the following corpus

```python
# ------------------------------------------------------------------ #
#                      Création de Bag of Words                      #
# ------------------------------------------------------------------ #

from sklearn.feature_extraction.text import CountVectorizer

corpus = [
    "I love NLP.",
    "NLP is amazing.",
    "I enjoy learning new things in NLP."
]

# Initialisation de CountVectorizer
bow_vect = CountVectorizer(max_features=1000) # most common 1000 words

# Fit and transform the corpus into a BoW representation
X = bow_vect.fit_transform(corpus)

print("Bag of Words:\n", X.toarray())
print("Vocabulary:", bow_vect.get_feature_names_out())
```

### **Exercise 4: TF-IDF**

Use the `TfidfVectorizer` from `scikit-learn` to create a TF-IDF representation of the same corpus. Store the result in `X_tfidf`

```python
# ------------------------------------------------------------------ #
#                               TF-IDF                               #
# ------------------------------------------------------------------ #

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfTransformer
import pandas as pd

corpus = ["I love NLP.", "NLP is amazing.", "I enjoy learning new things in NLP."]

# BoW to count the words
bow_vect = CountVectorizer(min_df=1) # min_df=1 : on garde tous les mots présents au moins 1 fois
term_freq_matrix = bow_vect.fit_transform(corpus) # Matrix BoW

# Initialisation transformeur TF-IDF
X_tfidf = TfidfTransformer(norm="l2") # norm="l2" : normalise les vecteurs pour éviter que les longues phrases dominent
X_tfidf.fit(term_freq_matrix) # Fit du TF-IDF
tf_idf_matrix = X_tfidf.transform(term_freq_matrix) # Transforme la matrice de fréq en pondération TF-IDF

# Checks
print("Vocabulary:", bow_vect.vocabulary_, '\n') # Affiche le vocalubary généré avec index
print(tf_idf_matrix.todense()) # Affiche TF-IDF matrix

# Data Viz => associe chaque mot à son poids IDF
df_idf = pd.DataFrame(X_tfidf.idf_, index=bow_vect.get_feature_names_out(), columns=["idf_weights"])
df_idf.sort_values(by=['idf_weights']) # Tri fréquence croissant 
```

### **Exercise 5: N-grams**

Generate `bigrams (2-grams)` from the corpus using `CountVectorizer`. Store the result in `X_bigram`

```python
# ------------------------------------------------------------------ #
#                         Création de Bigrams                        #
# ------------------------------------------------------------------ #

from sklearn.feature_extraction.text import CountVectorizer

corpus = ["I love NLP.", "NLP is amazing.", "I enjoy learning new things in NLP."]

# Initialisation de CountVectorizer
bigram_vectorizer = CountVectorizer(max_features=1000, ngram_range=(2, 2)) # most common 1000 words, ngram_range=(2, 2) from 1 mot seul to 2 mots

# Fit and transform the corpus into a Bigrams representation
X_bigram = bigram_vectorizer.fit_transform(corpus)

print("Bigrams:\n", X_bigram.toarray())
print("Bigrams Vocabulary:", bigram_vectorizer.get_feature_names_out())
```

## **4. Advanced Exercise: Custom Preprocessing Pipeline**

### **Exercise 6: Build a Custom Preprocessing Pipeline**

Combine all the preprocessing steps (tokenization, stop word removal, punctuation removal, stemming/lemmatization) into a single function. 

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer, PorterStemmer, SnowballStemmer

def text_preprocessing_pipeline(text: str, operation: str):
    """
    Preprocesses a given text by applying tokenization, stop word removal, punctuation removal, 
    and either stemming or lemmatization based on the specified operation.

    Parameters:
    ----------
    text : str
        The input text to preprocess.
    operation : str
        The operation to apply for word normalization. 
        Use "porter" for porter stemming, "snowball" for snowball stemming or "lemmatize" for lemmatization.

    Returns:
    -------
    list
        A list of preprocessed tokens after applying the specified operations.

    Steps:
    -----
    1. Tokenize the input text into words.
    2. Remove stop words using the NLTK stop words list.
    3. Remove punctuation and convert tokens to lowercase.
    4. Apply stemming or lemmatization based on the `ops` parameter.
    """

    # Step 1: Tokenize the text
    tokenized_text = word_tokenize(text=text, language="english", preserve_line=False)

    # Step 2: Remove stop words
    stop_words = set(stopwords.words('english'))
    filtered_tokens = [word for word in tokenized_text if word.lower() not in stop_words]

    # Step 3: Remove punctuation
    clean_tokenized_text = [word.lower() for word in filtered_tokens if word.isalpha()]

    # Step 4: Apply lemmatization
    if operation == "porter":
        stemmer = PorterStemmer()
        lemmatized_tokens = [stemmer.stem(word) for word in clean_tokenized_text]
    elif operation == "snowball":
        stemmer = SnowballStemmer()
        lemmatized_tokens = [stemmer.stem(word) for word in clean_tokenized_text]
    else:
        lemmatizer = WordNetLemmatizer()
        lemmatized_tokens = [lemmatizer.lemmatize(word) for word in clean_tokenized_text]
    
    return lemmatized_tokens
```

Apply this function to the following text

```python
text = "Natural Language Processing (NLP) is a fascinating field of study! It involves analyzing and understanding human language."
processed_text = text_preprocessing_pipeline(text, "lemmatize")
```

```python
print("Processed Text:", processed_text)
```
